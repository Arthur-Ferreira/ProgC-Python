### Exercício 4 - Classificador de Notas com Listas
Um professor quer automatizar a análise das notas de seus alunos. Você deve criar um programa que:

Orientações:
- Receba 4 notas digitadas pelo usuário (use input() e converta com float());
- Guarde as notas em uma lista;
- Calcule a média das notas (use um loop for para somar e depois divida por len());
- Use condicionais (if/elif/else) para classificar a média:
    - Aprovado se média >= 7;
    - Recuperação se média >= 5 e < 7;
    - Reprovado se média < 5.
- Use o operador in para verificar se alguma nota é igual a 10 e imprima uma mensagem de parabéns;
- Imprima a maior e a menor nota (use as funções max() e min()).

```python
aluno = input("Digite o nome do aluno: ")

notas = []
soma_notas = 0

while (len(notas) < 4):
    nota = float(input("Digite a nota do aluno: "))
    notas.append(nota)


for nota in notas:
    soma_notas += nota

media = soma_notas / len(notas)

print("--- Resultados --")
print(f"Média: {media:.2f}")


if media >= 7:
    print("Situação: Aprovado")
elif media >= 5 and media < 7:
    print("Situação: Recuperação")
else:
    print("Situação: Reprovado")

if 10 in notas:
    print("Parabéns! Você tirou 10 em alguma disciplina!")

print(f"Maior nota: {max(notas):.2f}")
print(f"Menor nota: {min(notas):.2f}")
```
